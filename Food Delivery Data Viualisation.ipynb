```python
# ================================================================
# FOOD DELIVERY BUSINESS PERFORMANCE
# COMPLETE VISUALIZATION PORTFOLIO
# Matplotlib + Seaborn
# ================================================================

# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Markdown

# Plotting configuration
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.titlesize"] = 15
plt.rcParams["axes.labelsize"] = 11

print("Libraries imported successfully!")


# ================================================================
# 2. LOAD DATASET
# ================================================================

# If using Google Colab/Jupyter in the same folder as the CSV
FILE_PATH = "Day14_Food_Delivery_Visualization_Dataset.csv"

df = pd.read_csv(FILE_PATH)

print("Dataset loaded successfully!")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")


# ================================================================
# 3. DATASET OVERVIEW
# ================================================================

print("\n" + "="*70)
print("DATASET OVERVIEW")
print("="*70)

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate records:")
print(df.duplicated().sum())


# ================================================================
# 4. DESCRIPTIVE STATISTICS
# ================================================================

print("\n" + "="*70)
print("DESCRIPTIVE STATISTICS")
print("="*70)

display(df.describe().T)


# ================================================================
# 5. DATA CLEANING & PREPROCESSING
# ================================================================

# Convert Date to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Convert numerical columns
numeric_columns = [
    "Orders",
    "Average_Order_Value",
    "Revenue",
    "Marketing_Spend",
    "Discounts",
    "Avg_Delivery_Minutes",
    "Customer_Rating",
    "Repeat_Customer_Percent"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove duplicate records
df = df.drop_duplicates()

# Remove records with invalid/missing dates
df = df.dropna(subset=["Date"])

# Create useful date columns
df["Year"] = df["Date"].dt.year
df["Month_Number"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Year_Month"] = df["Date"].dt.to_period("M").astype(str)

# Month ordering
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

df["Month_Name"] = pd.Categorical(
    df["Month_Name"],
    categories=month_order,
    ordered=True
)

print("\nCleaned dataset shape:")
print(df.shape)

print("\nMissing values after cleaning:")
display(df.isnull().sum().to_frame("Missing Values"))


# ================================================================
# 6. HELPER FUNCTION FOR INTERPRETATIONS
# ================================================================

def interpretation(text):
    display(Markdown("### Interpretation"))
    display(Markdown(text))


# ================================================================
# 7. LINE PLOT
# ORDERS OVER TIME
# ================================================================

daily_orders = (
    df.groupby("Date", as_index=False)["Orders"]
      .sum()
)

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=daily_orders,
    x="Date",
    y="Orders",
    marker="o"
)

plt.title("Daily Orders Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

highest_order_day = daily_orders.loc[
    daily_orders["Orders"].idxmax()
]

interpretation(
    f"The number of orders varies considerably over the year. "
    f"The highest recorded order volume occurred on "
    f"**{highest_order_day['Date'].strftime('%d %B %Y')}**, "
    f"with **{highest_order_day['Orders']:.0f} orders**. "
    f"The line plot helps identify peaks and fluctuations in demand over time."
)


# ================================================================
# 8. LINE PLOT
# MONTHLY REVENUE
# ================================================================

monthly_revenue = (
    df.groupby("Year_Month", as_index=False)["Revenue"]
      .sum()
)

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_revenue,
    x="Year_Month",
    y="Revenue",
    marker="o",
    linewidth=2
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

highest_revenue_month = monthly_revenue.loc[
    monthly_revenue["Revenue"].idxmax()
]

interpretation(
    f"Revenue changes across the months, indicating seasonal variation in "
    f"business performance. The highest monthly revenue was recorded in "
    f"**{highest_revenue_month['Year_Month']}**, reaching approximately "
    f"**{highest_revenue_month['Revenue']:,.2f}**."
)


# ================================================================
# 9. COMBINED LINE PLOT
# ORDERS AND REVENUE
# ================================================================

monthly_performance = (
    df.groupby("Year_Month", as_index=False)
      .agg(
          Orders=("Orders", "sum"),
          Revenue=("Revenue", "sum")
      )
)

fig, ax1 = plt.subplots(figsize=(14, 6))

ax1.plot(
    monthly_performance["Year_Month"],
    monthly_performance["Orders"],
    marker="o",
    linewidth=2,
    label="Orders"
)

ax1.set_xlabel("Month")
ax1.set_ylabel("Total Orders")
ax1.tick_params(axis="x", rotation=45)

ax2 = ax1.twinx()

ax2.plot(
    monthly_performance["Year_Month"],
    monthly_performance["Revenue"],
    marker="s",
    linewidth=2,
    label="Revenue"
)

ax2.set_ylabel("Total Revenue")

plt.title("Monthly Orders and Revenue Performance")

fig.tight_layout()
plt.show()

interpretation(
    "The combined plot allows order volume and revenue to be compared over "
    "the same time period. When both increase together, higher order volume "
    "is likely contributing to revenue growth. Differences between the two "
    "trends may indicate changes in average order value."
)


# ================================================================
# 10. BAR CHART
# REVENUE BY CITY
# ================================================================

city_performance = (
    df.groupby("City", as_index=False)
      .agg(
          Orders=("Orders", "sum"),
          Revenue=("Revenue", "sum")
      )
      .sort_values("Revenue", ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=city_performance,
    x="City",
    y="Revenue"
)

plt.title("Total Revenue by City")
plt.xlabel("City")
plt.ylabel("Total Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

top_city = city_performance.iloc[0]
bottom_city = city_performance.iloc[-1]

interpretation(
    f"**{top_city['City']}** generated the highest total revenue at "
    f"approximately **{top_city['Revenue']:,.2f}**, while "
    f"**{bottom_city['City']}** generated the lowest at "
    f"approximately **{bottom_city['Revenue']:,.2f}**. "
    f"This highlights differences in market performance across cities."
)


# ================================================================
# 11. BAR CHART
# ORDERS BY CITY
# ================================================================

plt.figure(figsize=(12, 6))

sns.barplot(
    data=city_performance.sort_values("Orders", ascending=False),
    x="City",
    y="Orders"
)

plt.title("Total Orders by City")
plt.xlabel("City")
plt.ylabel("Total Orders")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

top_order_city = city_performance.sort_values(
    "Orders", ascending=False
).iloc[0]

interpretation(
    f"**{top_order_city['City']}** recorded the highest total number of "
    f"orders, with approximately **{top_order_city['Orders']:,.0f} orders**. "
    f"Comparing this chart with the revenue chart helps determine whether "
    f"the strongest cities are driven primarily by order volume."
)


# ================================================================
# 12. BAR CHART
# REVENUE BY CUISINE
# ================================================================

cuisine_performance = (
    df.groupby("Cuisine", as_index=False)
      .agg(
          Orders=("Orders", "sum"),
          Revenue=("Revenue", "sum"),
          Average_Order_Value=("Average_Order_Value", "mean")
      )
      .sort_values("Revenue", ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=cuisine_performance,
    x="Cuisine",
    y="Revenue"
)

plt.title("Total Revenue by Cuisine")
plt.xlabel("Cuisine")
plt.ylabel("Total Revenue")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

top_cuisine = cuisine_performance.iloc[0]

interpretation(
    f"**{top_cuisine['Cuisine']}** generated the highest total revenue, "
    f"approximately **{top_cuisine['Revenue']:,.2f}**. "
    f"This suggests that this cuisine category is an important contributor "
    f"to overall business revenue."
)


# ================================================================
# 13. BAR CHART
# AVERAGE ORDER VALUE BY CUISINE
# ================================================================

plt.figure(figsize=(12, 6))

sns.barplot(
    data=cuisine_performance.sort_values(
        "Average_Order_Value",
        ascending=False
    ),
    x="Cuisine",
    y="Average_Order_Value"
)

plt.title("Average Order Value by Cuisine")
plt.xlabel("Cuisine")
plt.ylabel("Average Order Value")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

highest_aov_cuisine = cuisine_performance.sort_values(
    "Average_Order_Value",
    ascending=False
).iloc[0]

interpretation(
    f"The **{highest_aov_cuisine['Cuisine']}** category has the highest "
    f"average order value at approximately "
    f"**{highest_aov_cuisine['Average_Order_Value']:,.2f}**. "
    f"A high average order value can help a cuisine generate strong revenue "
    f"even without having the highest order count."
)


# ================================================================
# 14. SCATTER PLOT
# MARKETING SPEND VS REVENUE
# ================================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Marketing_Spend",
    y="Revenue",
    hue="City",
    style="Order_Channel",
    s=80
)

sns.regplot(
    data=df,
    x="Marketing_Spend",
    y="Revenue",
    scatter=False
)

plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")
plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)
plt.tight_layout()
plt.show()

marketing_correlation = df[
    "Marketing_Spend"
].corr(df["Revenue"])

interpretation(
    f"The Pearson correlation between marketing spend and revenue is "
    f"**{marketing_correlation:.2f}**. "
    f"The scatter plot shows whether records with greater marketing "
    f"investment generally correspond to higher revenue. However, "
    f"correlation does not by itself prove that marketing spend causes "
    f"higher revenue."
)


# ================================================================
# 15. SCATTER PLOT
# DELIVERY TIME VS CUSTOMER RATING
# ================================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Avg_Delivery_Minutes",
    y="Customer_Rating",
    hue="Weather",
    s=80
)

sns.regplot(
    data=df,
    x="Avg_Delivery_Minutes",
    y="Customer_Rating",
    scatter=False
)

plt.title("Delivery Time vs Customer Rating")
plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Customer Rating")
plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)
plt.tight_layout()
plt.show()

delivery_rating_correlation = df[
    "Avg_Delivery_Minutes"
].corr(df["Customer_Rating"])

interpretation(
    f"The Pearson correlation between delivery time and customer rating "
    f"is **{delivery_rating_correlation:.2f}**. "
    f"The relationship indicates whether longer delivery times tend to "
    f"coincide with higher or lower customer ratings."
)


# ================================================================
# 16. HISTOGRAM
# DISTRIBUTION OF ORDERS
# ================================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Orders",
    bins=20,
    kde=True
)

plt.title("Distribution of Orders")
plt.xlabel("Orders")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

interpretation(
    f"The average number of orders per record is "
    f"**{df['Orders'].mean():.2f}**, while the median is "
    f"**{df['Orders'].median():.2f}**. "
    f"The histogram shows the overall distribution and whether order "
    f"volume is concentrated around a typical range."
)


# ================================================================
# 17. HISTOGRAM
# DISTRIBUTION OF REVENUE
# ================================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Revenue",
    bins=25,
    kde=True
)

plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

interpretation(
    f"The mean revenue per record is approximately "
    f"**{df['Revenue'].mean():,.2f}**, compared with a median of "
    f"**{df['Revenue'].median():,.2f}**. "
    f"The shape of the distribution helps identify whether a small number "
    f"of high-revenue records may be influencing overall performance."
)


# ================================================================
# 18. HISTOGRAM
# CUSTOMER RATING DISTRIBUTION
# ================================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Customer_Rating",
    bins=15,
    kde=True
)

plt.title("Distribution of Customer Ratings")
plt.xlabel("Customer Rating")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

interpretation(
    f"Customer ratings have an average of "
    f"**{df['Customer_Rating'].mean():.2f}**. "
    f"The distribution indicates whether customer satisfaction is generally "
    f"consistent or spread across a wider range of ratings."
)


# ================================================================
# 19. BOX PLOT
# REVENUE BY CITY
# ================================================================

plt.figure(figsize=(13, 6))

sns.boxplot(
    data=df,
    x="City",
    y="Revenue"
)

plt.title("Revenue Distribution by City")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

interpretation(
    "The box plot compares the median, spread and potential outliers "
    "in revenue across cities. A larger box indicates greater variation "
    "in revenue, while isolated points may represent unusually high or "
    "low revenue observations."
)


# ================================================================
# 20. BOX PLOT
# DELIVERY TIME BY WEATHER
# ================================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Weather",
    y="Avg_Delivery_Minutes"
)

plt.title("Delivery Time Distribution by Weather")
plt.xlabel("Weather")
plt.ylabel("Average Delivery Time (Minutes)")
plt.tight_layout()
plt.show()

weather_delivery = (
    df.groupby("Weather")["Avg_Delivery_Minutes"]
      .mean()
      .sort_values(ascending=False)
)

slowest_weather = weather_delivery.index[0]

interpretation(
    f"Average delivery time is highest during **{slowest_weather}** "
    f"conditions, at approximately "
    f"**{weather_delivery.iloc[0]:.2f} minutes**. "
    f"This suggests that weather conditions may influence delivery operations."
)


# ================================================================
# 21. VIOLIN PLOT
# CUSTOMER RATING BY WEATHER
# ================================================================

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="Weather",
    y="Customer_Rating"
)

plt.title("Customer Rating Distribution by Weather")
plt.xlabel("Weather")
plt.ylabel("Customer Rating")
plt.tight_layout()
plt.show()

weather_rating = (
    df.groupby("Weather")["Customer_Rating"]
      .mean()
      .sort_values(ascending=False)
)

best_rating_weather = weather_rating.index[0]

interpretation(
    f"Customer ratings vary across weather conditions. "
    f"The highest average rating occurs during **{best_rating_weather}** "
    f"weather, at approximately **{weather_rating.iloc[0]:.2f}**. "
    f"The violin plot also shows the spread and density of ratings."
)


# ================================================================
# 22. COUNT PLOT
# ORDER CHANNEL
# ================================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Order_Channel"
)

plt.title("Order Records by Order Channel")
plt.xlabel("Order Channel")
plt.ylabel("Number of Records")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

channel_counts = df["Order_Channel"].value_counts()

top_channel = channel_counts.index[0]

interpretation(
    f"**{top_channel}** is the most frequently represented order channel, "
    f"with **{channel_counts.iloc[0]:,} records**. "
    f"This shows which channel has the strongest representation in the dataset."
)


# ================================================================
# 23. BAR CHART
# REVENUE BY ORDER CHANNEL
# ================================================================

channel_performance = (
    df.groupby("Order_Channel", as_index=False)
      .agg(
          Orders=("Orders", "sum"),
          Revenue=("Revenue", "sum"),
          Avg_Rating=("Customer_Rating", "mean")
      )
      .sort_values("Revenue", ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=channel_performance,
    x="Order_Channel",
    y="Revenue"
)

plt.title("Total Revenue by Order Channel")
plt.xlabel("Order Channel")
plt.ylabel("Total Revenue")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

best_channel = channel_performance.iloc[0]

interpretation(
    f"The **{best_channel['Order_Channel']}** channel generated the highest "
    f"total revenue at approximately **{best_channel['Revenue']:,.2f}**. "
    f"This can help identify which ordering platforms are most valuable "
    f"from a revenue perspective."
)


# ================================================================
# 24. COUNT PLOT
# WEATHER CONDITIONS
# ================================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Weather"
)

plt.title("Number of Records by Weather Condition")
plt.xlabel("Weather")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

common_weather = df["Weather"].value_counts().idxmax()

interpretation(
    f"**{common_weather}** is the most frequently observed weather "
    f"condition in the dataset. Weather frequency should be considered "
    f"when interpreting weather-based business performance."
)


# ================================================================
# 25. WEATHER VS BUSINESS PERFORMANCE
# ================================================================

weather_performance = (
    df.groupby("Weather", as_index=False)
      .agg(
          Average_Orders=("Orders", "mean"),
          Average_Revenue=("Revenue", "mean"),
          Average_Delivery_Time=("Avg_Delivery_Minutes", "mean"),
          Average_Rating=("Customer_Rating", "mean")
      )
      .sort_values("Average_Revenue", ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=weather_performance,
    x="Weather",
    y="Average_Revenue"
)

plt.title("Average Revenue by Weather Condition")
plt.xlabel("Weather")
plt.ylabel("Average Revenue")
plt.tight_layout()
plt.show()

best_weather_revenue = weather_performance.iloc[0]

interpretation(
    f"**{best_weather_revenue['Weather']}** conditions have the highest "
    f"average revenue at approximately "
    f"**{best_weather_revenue['Average_Revenue']:,.2f}**. "
    f"This suggests that weather may be associated with differences "
    f"in food delivery business performance."
)


# ================================================================
# 26. ORDER CHANNEL VS CUSTOMER RATING
# ================================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Order_Channel",
    y="Customer_Rating"
)

plt.title("Customer Ratings by Order Channel")
plt.xlabel("Order Channel")
plt.ylabel("Customer Rating")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

interpretation(
    "The box plot compares customer rating distributions across ordering "
    "channels. Differences in median ratings and spread can indicate whether "
    "customer satisfaction varies depending on the platform used to place orders."
)


# ================================================================
# 27. CITY VS CUISINE REVENUE
# HEATMAP
# ================================================================

city_cuisine = pd.pivot_table(
    df,
    values="Revenue",
    index="City",
    columns="Cuisine",
    aggfunc="sum"
)

plt.figure(figsize=(14, 7))

sns.heatmap(
    city_cuisine,
    annot=True,
    fmt=".0f",
    linewidths=0.5
)

plt.title("Revenue Heatmap: City vs Cuisine")
plt.xlabel("Cuisine")
plt.ylabel("City")
plt.tight_layout()
plt.show()

interpretation(
    "The heatmap shows which city-cuisine combinations generate the most "
    "revenue. Darker or larger-valued cells indicate stronger revenue "
    "contributions, allowing high-performing combinations to be identified quickly."
)


# ================================================================
# 28. CORRELATION HEATMAP
# ================================================================

correlation_columns = [
    "Orders",
    "Average_Order_Value",
    "Revenue",
    "Marketing_Spend",
    "Discounts",
    "Avg_Delivery_Minutes",
    "Customer_Rating",
    "Repeat_Customer_Percent"
]

correlation_matrix = df[correlation_columns].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Heatmap of Business Performance Variables")
plt.tight_layout()
plt.show()

interpretation(
    "The correlation heatmap summarizes relationships among the key numerical "
    "variables. Strong positive correlations indicate variables that tend to "
    "increase together, while negative correlations indicate an inverse "
    "relationship. Correlation should be interpreted as association rather "
    "than proof of causation."
)


# ================================================================
# 29. TOP BUSINESS PERFORMANCE METRICS
# ================================================================

total_orders = df["Orders"].sum()
total_revenue = df["Revenue"].sum()
total_marketing = df["Marketing_Spend"].sum()
average_order_value = df["Average_Order_Value"].mean()
average_delivery = df["Avg_Delivery_Minutes"].mean()
average_rating = df["Customer_Rating"].mean()
average_repeat_customer = df["Repeat_Customer_Percent"].mean()

print("\n" + "="*70)
print("OVERALL BUSINESS PERFORMANCE")
print("="*70)

print(f"Total Orders              : {total_orders:,.0f}")
print(f"Total Revenue             : {total_revenue:,.2f}")
print(f"Total Marketing Spend     : {total_marketing:,.2f}")
print(f"Average Order Value       : {average_order_value:,.2f}")
print(f"Average Delivery Time     : {average_delivery:.2f} minutes")
print(f"Average Customer Rating   : {average_rating:.2f}")
print(f"Average Repeat Customer % : {average_repeat_customer:.2f}%")


# ================================================================
# 30. AUTOMATIC KEY FINDINGS
# ================================================================

top_city = city_performance.iloc[0]["City"]
top_city_revenue = city_performance.iloc[0]["Revenue"]

top_cuisine = cuisine_performance.iloc[0]["Cuisine"]
top_cuisine_revenue = cuisine_performance.iloc[0]["Revenue"]

top_channel = channel_performance.iloc[0]["Order_Channel"]
top_channel_revenue = channel_performance.iloc[0]["Revenue"]

best_weather = weather_performance.iloc[0]["Weather"]
best_weather_revenue = weather_performance.iloc[0]["Average_Revenue"]

highest_delivery_weather = weather_delivery.index[0]
highest_delivery_time = weather_delivery.iloc[0]


# ================================================================
# 31. FINAL SUMMARY
# ================================================================

display(Markdown("# Final Summary and Key Findings"))

summary_points = [
    f"**Overall business:** The dataset contains {len(df):,} records with "
    f"approximately **{total_orders:,.0f} total orders** and "
    f"**{total_revenue:,.2f} total revenue**.",

    f"**City performance:** {top_city} is the strongest city by total revenue, "
    f"generating approximately **{top_city_revenue:,.2f}**.",

    f"**Cuisine performance:** {top_cuisine} is the highest-revenue cuisine, "
    f"with approximately **{top_cuisine_revenue:,.2f}** in total revenue.",

    f"**Order channel:** {top_channel} generated the highest total revenue "
    f"among the available ordering channels, at approximately "
    f"**{top_channel_revenue:,.2f}**.",

    f"**Marketing and revenue:** Marketing spend has a Pearson correlation "
    f"of **{marketing_correlation:.2f}** with revenue, showing the direction "
    f"and strength of their linear association.",

    f"**Delivery and ratings:** Delivery time has a Pearson correlation of "
    f"**{delivery_rating_correlation:.2f}** with customer rating.",

    f"**Weather:** {best_weather} weather has the highest average revenue "
    f"among the weather categories at approximately "
    f"**{best_weather_revenue:,.2f}**.",

    f"**Delivery operations:** {highest_delivery_weather} conditions have "
    f"the highest average delivery time at approximately "
    f"**{highest_delivery_time:.2f} minutes**.",

    f"**Customer satisfaction:** The overall average customer rating is "
    f"approximately **{average_rating:.2f}/5**.",

    f"**Repeat customers:** The average repeat-customer percentage is "
    f"approximately **{average_repeat_customer:.2f}%**."
]

for point in summary_points:
    display(Markdown("- " + point))


# ================================================================
# 32. FINAL CONCLUSION
# ================================================================

display(Markdown("""
## Conclusion

The visualization portfolio provides a comprehensive view of food delivery
business performance using Matplotlib and Seaborn. The analysis examined
time-based trends, city and cuisine performance, marketing investment,
delivery operations, customer satisfaction, weather conditions, and ordering
channels.

The visualizations demonstrate that business performance is influenced by
multiple dimensions rather than a single factor. Revenue and order volume
show variation over time, while cities and cuisines differ substantially in
their contribution to overall revenue. The marketing-spend analysis provides
an indication of the relationship between promotional investment and revenue,
while the delivery-time analysis helps assess the connection between
operational performance and customer satisfaction.

Weather and order-channel visualizations further highlight how external
conditions and customer ordering behavior can affect business outcomes.

Overall, the analysis can help a food delivery business identify high-value
markets, strong cuisine categories, effective ordering channels, operational
challenges, and areas where customer experience can potentially be improved.
"""))

print("\n" + "="*70)
print("VISUALIZATION PORTFOLIO COMPLETED SUCCESSFULLY!")
print("="*70)
```
